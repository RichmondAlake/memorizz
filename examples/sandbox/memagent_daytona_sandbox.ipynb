# MemAgent with Daytona Sandbox — Full Dev Environments in the Cloud

This notebook demonstrates how to use **MemAgent** with the **Daytona** sandbox provider. Daytona provides full cloud development environments with unlimited runtime, GPU support, and Git integration.

## Why Daytona?

While E2B is optimized for quick AI agent code execution, Daytona shines when you need:
- **Unlimited session duration** — no 24-hour limit
- **GPU access** — run ML workloads inside the sandbox
- **Full dev environments** — Git, package managers, configurable CPU/memory/disk
- **~90ms cold starts** — the fastest cold start of any cloud sandbox

```python
import os
import getpass

def set_env_securely(var_name, prompt):
    """Securely prompt for and set an environment variable."""
    value = getpass.getpass(prompt)
    os.environ[var_name] = value

set_env_securely("DAYTONA_API_KEY", "Enter your Daytona API key: ")
set_env_securely("OPENAI_API_KEY", "Enter your OpenAI API key: ")
```

## Step 3: Create a MemAgent with Daytona Sandbox

### Option A: Simple — just pass the provider name

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    force=True
)
```

```python
from memorizz.memagent.core import MemAgent
from memorizz.llms.llm_factory import create_llm_provider

llm = create_llm_provider({
    "provider": "openai",
    "model": "gpt-4o",
    "api_key": os.getenv("OPENAI_API_KEY"),
})

# Simple: just pass the provider name
agent = MemAgent(
    model=llm,
    sandbox_provider="daytona",
    instruction=(
        "You are a Python developer assistant. "
        "When asked to solve problems, write Python code and execute it "
        "using the execute_code tool. Always explain your approach."
    ),
)

print(f"Sandbox provider: {agent.get_sandbox_provider_name()}")
print(f"Has sandbox: {agent.has_sandbox()}")
```

## Step 4: Code Generation and Execution

The agent works identically regardless of which sandbox provider you use. The interface is the same — only the execution backend changes.

### Example 1: Data Processing

```python
response = agent.run(
    "Create a Python dictionary of the 10 largest countries by area, "
    "then sort them by area in descending order and print a formatted table."
)
print(f"\nAgent: {response}")
```

### Example 2: Algorithm Implementation

```python
response = agent.run(
    "Implement a binary search algorithm in Python. "
    "Test it by searching for the number 42 in a sorted list of 100 elements. "
    "Print each step of the search to show how the algorithm narrows down the range."
)
print(f"\nAgent: {response}")
```

### Example 3: Working with JSON Data

```python
response = agent.run(
    "Create a sample JSON dataset of 5 fictional employees with fields: "
    "name, department, salary, years_experience. "
    "Then write code to: \n"
    "1. Find the highest paid employee\n"
    "2. Calculate average salary by department\n"
    "3. Find employees with more than 5 years experience"
)
print(f"\nAgent: {response}")
```

## Key Takeaways

1. **Same API, different backend** — Switching from E2B to Daytona is just changing `"e2b"` to `"daytona"`. The tool interface is identical.
2. **Unlimited runtime** — Daytona sandboxes have no session time limit, making them ideal for long-running computations.
3. **GPU support** — Daytona supports GPU allocation for ML workloads.
4. **Full dev environments** — Daytona sandboxes include Git, package managers, and configurable resources.

## When to Use Daytona vs E2B

| Use Case | Recommended Provider |
|----------|---------------------|
| Quick computations, data analysis | E2B (simpler, AI-optimized) |
| Long-running ML training | Daytona (unlimited runtime, GPU) |
| Full development workflows | Daytona (Git, package managers) |
| Prototyping / getting started | E2B (easiest setup) |
